```python
print("==========SISTEMAS DE FUNCIONARIOS=============")

funcionarios = ["joao","gabriel","rodrigues"]
salarios = [1500.00,1621.99,1233.33]
while True:
  print("Confira o menu abaixo")
  print("Opções:\n 1-cadastrar funcionario\n 2-ver resultados\n 3-Mostrar salarios e os funcionarios\n 0-encerramento")


  opcao = int(input("Escolha uma opção:"))
  if opcao == 1:
    adicionar = input("Digite o nome do Funcionario:")
    adicionar2 = float(input("Digite o salario"))
    funcionarios.append(adicionar)
    salarios.append(adicionar2)

    print("Cadastro realizado com sucesso")

    for i,j in zip(funcionarios,salarios):
      print(i,j)


  elif opcao == 2:
    total_salarios = 0

    for i in salarios:
      total_salarios = total_salarios + i
      media = total_salarios / len(salarios)

    print(f"O resultado é {media}")

  elif opcao == 3:
    for i,j in zip(funcionarios,salarios):
      print(f"aqui está os funcionarios e seus salarios: {i},{j}")
  break

print("Sistema encerrado")







































```
